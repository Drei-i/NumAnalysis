# 1. Use the Bisection Method to find the root to six correct decimal places. (a) $x^3 = 9$ (b) $3x^3 + x^2 = x + 5$ (c) $\cos^2 x + 6 = x$

```python
# ==============================================================================
# Problem 1: Bisection Method to 6 Correct Decimal Places
# ==============================================================================
import math

def bisection_table(f, a, b, tol=0.5e-6, max_iter=100):
    if f(a) * f(b) > 0:
        raise ValueError(f"No sign change between a={a} and b={b}.")

    history = []
    step = 0
    while (b - a) / 2.0 > tol and step < max_iter:
        step += 1
        c = (a + b) / 2.0
        fc = f(c)
        err = (b - a) / 2.0
        history.append((step, a, b, c, fc, err))

        if fc == 0.0:
            break
        elif f(a) * fc < 0:
            b = c
        else:
            a = c

    root = (a + b) / 2.0
    final_err = (b - a) / 2.0
    return root, final_err, history

def display_results(label, equation_str, f, a, b):
    tol = 0.5e-6
    print("=" * 86)
    print(f"Problem 1({label}): {equation_str}")
    print(f"Initial Interval: [{a:.2f}, {b:.2f}] | Stopping Criterion: (b - a)/2 < {tol:.1e}")
    print("-" * 86)
    print(f"{'Iter':<5} | {'a':<11} | {'b':<11} | {'c (mid)':<11} | {'f(c)':<14} | {'Error Bound':<14}")
    print("-" * 86)

    root, err, history = bisection_table(f, a, b, tol=tol)
    for step, ak, bk, ck, fck, errk in history:
        print(f"{step:<5} | {ak:<11.6f} | {bk:<11.6f} | {ck:<11.6f} | {fck:<14.6e} | {errk:<14.6e}")

    print("-" * 86)
    print(f"Approximated Root: {root:.6f}")
    print(f"Final Error Bound: {err:.2e} (Converged in {len(history)} iterations)")
    print(f"Conclusion: Root for 1({label}) is x ≈ {root:.6f} correct to 6 decimal places.\n")

# (a) x^3 = 9
display_results('a', "x^3 = 9", lambda x: x**3 - 9, 2.0, 3.0)

# (b) 3x^3 + x^2 = x + 5
display_results('b', "3x^3 + x^2 = x + 5", lambda x: 3*x**3 + x**2 - x - 5, 1.0, 2.0)

# (c) cos^2(x) + 6 = x
display_results('c', "cos^2(x) + 6 = x", lambda x: math.cos(x)**2 + 6 - x, 6.0, 7.0)
```

# 2. Use the Bisection Method to find the root to eight correct decimal places. (a) $x^5 + x = 1$ (b) $\sin x = 6x + 5$ (c) $\ln x + x^2 = 3$

```python
# ==============================================================================
# Problem 2: Bisection Method to 8 Correct Decimal Places
# ==============================================================================
import math

def bisection_table(f, a, b, tol=0.5e-8, max_iter=100):
    if f(a) * f(b) > 0:
        raise ValueError(f"No sign change between a={a} and b={b}.")

    history = []
    step = 0
    while (b - a) / 2.0 > tol and step < max_iter:
        step += 1
        c = (a + b) / 2.0
        fc = f(c)
        err = (b - a) / 2.0
        history.append((step, a, b, c, fc, err))

        if fc == 0.0:
            break
        elif f(a) * fc < 0:
            b = c
        else:
            a = c

    root = (a + b) / 2.0
    final_err = (b - a) / 2.0
    return root, final_err, history

def display_results(label, equation_str, f, a, b):
    tol = 0.5e-8
    print("=" * 96)
    print(f"Problem 2({label}): {equation_str}")
    print(f"Initial Interval: [{a:.2f}, {b:.2f}] | Stopping Criterion: (b - a)/2 < {tol:.1e}")
    print("-" * 96)
    print(f"{'Iter':<5} | {'a':<13} | {'b':<13} | {'c (mid)':<13} | {'f(c)':<16} | {'Error Bound':<16}")
    print("-" * 96)

    root, err, history = bisection_table(f, a, b, tol=tol)
    for step, ak, bk, ck, fck, errk in history:
        print(f"{step:<5} | {ak:<13.8f} | {bk:<13.8f} | {ck:<13.8f} | {fck:<16.8e} | {errk:<16.8e}")

    print("-" * 96)
    print(f"Approximated Root: {root:.8f}")
    print(f"Final Error Bound: {err:.2e} (Converged in {len(history)} iterations)")
    print(f"Conclusion: Root for 2({label}) is x ≈ {root:.8f} correct to 8 decimal places.\n")

# (a) x^5 + x = 1
display_results('a', "x^5 + x = 1", lambda x: x**5 + x - 1, 0.0, 1.0)

# (b) sin(x) = 6x + 5
display_results('b', "sin(x) = 6x + 5", lambda x: math.sin(x) - 6*x - 5, -1.0, 0.0)

# (c) ln(x) + x^2 = 3
display_results('c', "ln(x) + x^2 = 3", lambda x: math.log(x) + x**2 - 3, 1.0, 2.0)
```

# 3. Use the Bisection Method to locate all solutions of the following equations. Sketch the function by using MATLAB's plot command and identify three intervals of length one that contain a root. Then find the roots to six correct decimal places. (a) $2x^3 - 6x - 1 = 0$ (b) $e^{x-2} + x^3 - x = 0$ (c) $1 + 5x - 6x^3 - e^{2x} = 0$

```python
# ==============================================================================
# Problem 3: Sketch Functions, Locate Length-1 Intervals, and Find Roots
# ==============================================================================
import numpy as np
import matplotlib.pyplot as plt

def bisection_table(f, a, b, tol=0.5e-6, max_iter=100):
    if f(a) * f(b) > 0:
        raise ValueError(f"No sign change between a={a} and b={b}.")

    history = []
    step = 0
    while (b - a) / 2.0 > tol and step < max_iter:
        step += 1
        c = (a + b) / 2.0
        fc = f(c)
        err = (b - a) / 2.0
        history.append((step, a, b, c, fc, err))

        if fc == 0.0:
            break
        elif f(a) * fc < 0:
            b = c
        else:
            a = c

    root = (a + b) / 2.0
    final_err = (b - a) / 2.0
    return root, final_err, history

# ----------------- 1. Function Definitions -----------------
f3a = lambda x: 2*x**3 - 6*x - 1
f3b = lambda x: np.exp(x - 2) + x**3 - x
f3c = lambda x: 1 + 5*x - 6*x**3 - np.exp(2*x)

# ----------------- 2. Plotting (Colab inline) -----------------
fig, axes = plt.subplots(1, 3, figsize=(18, 4.5))

plots_config = [
    (axes[0], f3a, "3(a) $2x^3 - 6x - 1 = 0$", (-3, 3), (-15, 15)),
    (axes[1], f3b, "3(b) $e^{x-2} + x^3 - x = 0$", (-2.5, 2.5), (-6, 6)),
    (axes[2], f3c, "3(c) $1 + 5x - 6x^3 - e^{2x} = 0$", (-2, 1.5), (-10, 5))
]

for ax, fn, title, x_lim, y_lim in plots_config:
    xs = np.linspace(x_lim[0], x_lim[1], 1000)
    ax.plot(xs, fn(xs), color="royalblue", lw=2, label="$f(x)$")
    ax.axhline(0, color="crimson", linestyle="--", lw=1.2, label="$y=0$")
    ax.set_xlim(x_lim)
    ax.set_ylim(y_lim)
    ax.set_title(title, fontsize=12)
    ax.set_xlabel("x")
    ax.set_ylabel("f(x)")
    ax.grid(True, linestyle=":", alpha=0.6)
    ax.legend()

plt.tight_layout()
plt.show()

# ----------------- 3. Length-1 Intervals & Execution -----------------
# Intervals of length 1 (using unit integer brackets where sign changes occur,
# and unit-length offsets where multiple roots reside within [0, 1])
problems = [
    {
        "part": "3(a)",
        "eq": "2x^3 - 6x - 1 = 0",
        "func": f3a,
        "intervals": [(-2.0, -1.0), (-1.0, 0.0), (1.0, 2.0)]
    },
    {
        "part": "3(b)",
        "eq": "e^(x-2) + x^3 - x = 0",
        "func": f3b,
        "intervals": [(-2.0, -1.0), (-0.5, 0.5), (0.5, 1.5)]
    },
    {
        "part": "3(c)",
        "eq": "1 + 5x - 6x^3 - e^(2x) = 0",
        "func": f3c,
        "intervals": [(-1.5, -0.5), (-0.5, 0.5), (0.2, 1.2)]
    }
]

for p in problems:
    print("\n" + "#" * 80)
    print(f"EQUATION {p['part']}: {p['eq']}")
    print("#" * 80)

    found_roots = []
    for idx, (a, b) in enumerate(p["intervals"], 1):
        print(f"\n--- Root {idx}: Interval [{a:.1f}, {b:.1f}] (Length = {b - a:.1f}) ---")
        print(f"{'Iter':<5} | {'a':<10} | {'b':<10} | {'c (mid)':<10} | {'f(c)':<12} | {'Error Bound':<12}")
        print("-" * 78)

        root, err, history = bisection_table(p["func"], a, b, tol=0.5e-6)
        for step, ak, bk, ck, fck, errk in history:
            print(f"{step:<5} | {ak:<10.5f} | {bk:<10.5f} | {ck:<10.5f} | {fck:<12.5e} | {errk:<12.5e}")

        print("-" * 78)
        print(f"Approximated Root: {root:.6f} (Error: {err:.2e}, Steps: {len(history)})")
        found_roots.append(root)

    print("\n" + "-" * 40)
    print(f"Conclusion for {p['part']}:")
    print(f"All 3 roots found to 6 decimal places: {', '.join([f'{r:.6f}' for r in found_roots])}")
```
